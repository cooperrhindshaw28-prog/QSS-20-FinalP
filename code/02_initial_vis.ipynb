```python
# QSS 20 Final Project
# Cooper Hindshaw



import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_stata("../dataverse_files/UCT_FINAL_CLEAN.dta")


treatment_counts = df["treat"].value_counts().sort_index()
```

```python
treatment_counts.plot(kind="bar")

plt.xlabel("Cash Transfer Treatment Status")
plt.ylabel("Number of Observations")
plt.title("Number of Observations by Cash Transfer Treatment Status")
plt.xticks([0, 1], ["Control", "Received Cash Transfer"], rotation=0)

plt.savefig("../output/cash_transfer_treatment_counts.png",
            bbox_inches="tight")


plt.show()

```

```python
df["asset_total_ppp0"].plot(
    kind="hist",
    bins=30)

plt.xlabel("Baseline Household Assets (PPP)")
plt.ylabel("Number of Observations")
plt.title("Distribution of Baseline Household Assets")

plt.savefig("../output/baseline_asset_distribution.png",
            bbox_inches="tight")

plt.show()
```

```python
plt.scatter(
    df["asset_total_ppp0"],
    df["asset_total_ppp1"],
    alpha=0.3
)

plt.xlabel("Baseline Household Assets (PPP)")
plt.ylabel("Follow-Up Household Assets (PPP)")
plt.title("Relationship Between Baseline and Follow-Up Household Assets")

plt.savefig("../output/baseline_vs_followup_assets.png",
            bbox_inches="tight")

plt.show()
```
